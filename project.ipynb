```python

import cv2
import torch
from ultralytics import YOLO

# Select GPU if available, otherwise use CPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load YOLOv8 model
model = YOLO("yolov8x.pt").to(device)

# Open webcam
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print("Error: Could not open webcam.")
    exit()

# Get all YOLO class labels
ALL_CLASSES = set(model.names.values())

while True:
    ret, frame = cap.read()
    if not ret:
        print("Error: Could not read frame.")
        break

    # Convert frame to RGB for YOLO
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Perform object detection
    results = model.predict(frame_rgb, device=device, conf=0.6, iou=0.5, verbose=False)

    detected_animals = []

    for result in results:
        if hasattr(result, "boxes"):
            for box in result.boxes:
                class_id = int(box.cls.item())
                confidence = float(box.conf.item())

                if class_id in model.names:
                    label = model.names[class_id]
                    print(f"Detected: {label} (Confidence: {confidence:.2f})")

                    # Check if the detected object is an animal
                    if label in ALL_CLASSES and confidence > 0.6:
                        detected_animals.append(label)

                        # Get bounding box coordinates
                        x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())

                        # Draw bounding box
                        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 3)

                        # Draw label text
                        cv2.putText(frame, f"{label} {confidence:.2f}", 
                                    (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 
                                    0.8, (0, 255, 0), 2)

    if detected_animals:
        print(f"Detected Animals: {', '.join(set(detected_animals))}")

    # Show the output
    cv2.imshow("Animal Detector", frame)

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
